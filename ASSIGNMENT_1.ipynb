```python

from decimal import Decimal, ROUND_HALF_UP

# Seasonal rate details
SEASONAL_BASE_RATE = Decimal("32.00")
SEASONAL_FAT_STEP_VALUE = Decimal("0.45")
BONUS_RATE = Decimal("0.50")

def get_rate_card(month, base_rate, fat_step_value):
    if 4 <= month <= 6:
        return SEASONAL_BASE_RATE, SEASONAL_FAT_STEP_VALUE
    return Decimal(str(base_rate)), Decimal(str(fat_step_value))


def calculate_bonus(total_litres, litres):
    if total_litres > 500:
        return Decimal(str(litres)) * BONUS_RATE
    return Decimal("0.00")


def calculate_payment(fat, snf, litres, total_litres,
                      month, reference_fat, fat_step_size,
                      original_base_rate, original_step_value):

    fat = Decimal(str(fat))
    snf = Decimal(str(snf))
    litres = Decimal(str(litres))

    base_rate, step_value = get_rate_card(
        month, original_base_rate, original_step_value
    )

    fat_steps = (fat - Decimal(str(reference_fat))) / Decimal(str(fat_step_size))

    milk_rate = base_rate + (fat_steps * step_value)

    milk_amount = milk_rate * litres

    bonus = calculate_bonus(total_litres, litres)

    total_payment = milk_amount + bonus

    return {
        "FAT": fat,
        "SNF": snf,
        "Milk Rate": milk_rate,
        "Milk Amount": milk_amount.quantize(
            Decimal("0.01"), rounding=ROUND_HALF_UP
        ),
        "Bonus": bonus.quantize(
            Decimal("0.01"), rounding=ROUND_HALF_UP
        ),
        "Total Payment": total_payment.quantize(
            Decimal("0.01"), rounding=ROUND_HALF_UP
        )
    }


# Example calculation
result = calculate_payment(
    fat=3.50,
    snf=8.50,
    litres=550,
    total_litres=550,
    month=5,
    reference_fat=3.00,
    fat_step_size=0.10,
    original_base_rate=30.00,
    original_step_value=0.40
)

for key, value in result.items():
    print(key, ":", value)
```
