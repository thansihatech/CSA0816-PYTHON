```python

import time

# Original program using nested loops
def original_program(records):
    totals = {}
    comparisons = 0

    farmer_ids = set(record[0] for record in records)

    for farmer in farmer_ids:
        total = 0

        for record in records:
            comparisons += 1

            if record[0] == farmer:
                total += record[1]

        totals[farmer] = total

    return totals, comparisons


# Redesigned program using dictionary
def redesigned_program(records):
    totals = {}
    operations = 0

    for farmer, litres in records:
        totals[farmer] = totals.get(farmer, 0) + litres
        operations += 1

    return totals, operations


# Generate sample milk records
records = []

for i in range(5000):
    farmer_id = "F" + str(i % 1000)
    litres = (i % 20) + 1
    records.append((farmer_id, litres))


# Test original program
start = time.perf_counter()
old_result, old_count = original_program(records)
old_time = time.perf_counter() - start


# Test redesigned program
start = time.perf_counter()
new_result, new_count = redesigned_program(records)
new_time = time.perf_counter() - start


# Validate both results
print("Original Result:", old_result)
print("Redesigned Result:", new_result)

print("\nValidation:",
      "PASS" if old_result == new_result else "FAIL")

print("\nPerformance Comparison")
print("Number of Records:", len(records))

print("Original Comparisons:", old_count)
print("Redesigned Operations:", new_count)

print("Original Time:", round(old_time, 6), "seconds")
print("Redesigned Time:", round(new_time, 6), "seconds")
```
